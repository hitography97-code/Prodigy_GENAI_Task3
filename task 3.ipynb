```python
import random
```

```python
text = """
Artificial Intelligence is transforming industries.
Data Science helps in decision making.
GPT-2 can generate human-like text.
Machine learning models learn from data.
Cloud computing enables scalable solutions.
Visualization helps in understanding data patterns.
"""
```

```python
words = text.split()
```

```python
markov_chain = {}
for i in range(len(words)-1):
    curr_word = words[i]
    next_word = words[i+1]
    if curr_word not in markov_chain:
        markov_chain[curr_word] = []
    markov_chain[curr_word].append(next_word)
```

```python
def generate_text(chain, start_word, length=20):
    word = start_word
    output = [word]
    for _ in range(length):
        if word in chain:
            word = random.choice(chain[word])
            output.append(word)
        else:
            break
    return " ".join(output)
```

```python
print("Generated Text:\n")
print(generate_text(markov_chain, "Artificial", 30))
```
